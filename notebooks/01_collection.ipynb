```python
def collect_search_page(query, page=0):
    url = "https://rabota.by/search/vacancy"

    params = {
        "text": query,
        "page": page
    }

    response = requests.get(
        url,
        params=params,
        headers=headers,
        timeout=20
    )

    if response.status_code != 200:
        print(
            f"Ошибка для {query}, страница {page}:",
            response.status_code
        )
        return []

    soup = BeautifulSoup(
        response.text,
        "html.parser"
    )

    cards = soup.find_all(
        "article",
        attrs={"data-qa": "vacancy-serp__vacancy"}
    )

    jobs = []

    for card in cards:

        title_tag = card.find(
            "a",
            attrs={"data-qa": "serp-item__title"}
        )

        company_tag = card.find(
            attrs={
                "data-qa": "vacancy-serp__vacancy-employer"
            }
        )

        address_tag = card.find(
            attrs={
                "data-qa": "vacancy-serp__vacancy-address"
            }
        )

        experience_tag = card.select_one(
            '[data-qa^="vacancy-serp__vacancy-work-experience"]'
        )

        title = (
            title_tag.get_text(" ", strip=True)
            if title_tag
            else None
        )

        vacancy_url = (
            title_tag.get("href")
            if title_tag
            else None
        )

        company = (
            company_tag.get_text(" ", strip=True)
            if company_tag
            else None
        )

        address = (
            address_tag.get_text(" ", strip=True)
            if address_tag
            else None
        )

        experience = (
            experience_tag.get_text(" ", strip=True)
            if experience_tag
            else None
        )

        job = {
            "title": title,
            "company": company,
            "address": address,
            "experience": experience,
            "url": vacancy_url,
            "query": query,
            "page": page
        }

        jobs.append(job)

    return jobs
```

```python
# начало встречаться часто r и другие буквы в скиллах, надо регулярные выражения добавить
import re
def extract_skills(description):

    if description is None:
        return []

    text = description.lower()

    found_skills = []

    for skill, variants in SKILLS.items():

        for variant in variants:

            pattern = (#непосредственно до и после найденной технологии не должно быть буквы, цифры
                r"(?<!\w)"
                + re.escape(variant.lower().strip())
                + r"(?!\w)"
            )

            if re.search(pattern, text):
                found_skills.append(skill)
                break

    return found_skills
```

```python
def extract_skills(description):

    if not isinstance(description, str):
        return []

    if not description.strip():
        return []

    text = description.lower()

    found_skills = []

    for skill, variants in SKILLS.items():

        for variant in variants:

            pattern = (
                r"(?<!\w)"
                + re.escape(variant.lower().strip())
                + r"(?!\w)"
            )

            if re.search(pattern, text):
                found_skills.append(skill)
                break

    return found_skills
```

```python
df["skills"] = (
    df["description"]
    .apply(extract_skills)
)
```

```python
def show_skill_context(description, term, context=80):

    if not isinstance(description, str):
        return None

    text = description.lower()
    position = text.find(term.lower())

    if position == -1:
        return None

    start = max(0, position - context)
    end = min(len(description), position + len(term) + context)

    return description[start:end]




example = df[df["rag"] == 1].iloc[0]

print(
    show_skill_context(
        example["description"],
        "rag"
    )
)
```
