```python
def classify_role(title):

    if not isinstance(title, str):
        return "Other"

    title = title.lower()

    # MLOps
    if re.search(r"\bmlops\b", title):
        return "MLOps Engineer"

    # ML Engineer
    if (
        "machine learning engineer" in title
        or "ml engineer" in title
        or "machine learning developer" in title
        or "ml developer" in title
        or "инженер машинного обучения" in title
        or "ml разработчик" in title
    ):
        return "ML Engineer"

    # Data Scientist
    if (
        "data scientist" in title
        or "дата сайентист" in title
        or "специалист по data science" in title
    ):
        return "Data Scientist"

    # Data Engineer
    if (
        "data engineer" in title
        or "инженер данных" in title
    ):
        return "Data Engineer"

    # BI Analyst
    if (
        "bi analyst" in title
        or "bi-analyst" in title
        or "bi аналитик" in title
        or "bi-аналитик" in title
        or "business intelligence analyst" in title
    ):
        return "BI Analyst"

    # Data Analyst
    if (
        "data analyst" in title
        or "аналитик данных" in title
    ):
        return "Data Analyst"

    # AI Engineer
    if (
        "ai engineer" in title
        or "ai developer" in title
        or "llm engineer" in title
        or "genai engineer" in title
        or "generative ai engineer" in title
        or "инженер искусственного интеллекта" in title
    ):
        return "AI Engineer"

    # Python Developer
    if (
        "python developer" in title
        or "python разработчик" in title
        or "python-разработчик" in title
        or "разработчик python" in title
    ):
        return "Python Developer"

    return "Other"
```

```python
df["role"] = (
    df["title"]
    .apply(classify_role)
)
```

```python
ROLE_PATTERNS = {

    "MLOps Engineer": [
        r"\bmlops\b",
        r"\bml[\s-]?ops\b",
        r"machine learning operations"
    ],

    "AI Engineer": [
    r"\bai engineer\b",
    r"\bai[\s-]?engineer\b",
    r"\bai developer\b",
    r"\bai[\s/-]?ml engineer\b",
    r"\bllm engineer\b",
    r"\bllm developer\b",
    r"\bgen[\s-]?ai engineer\b",
    r"\bgenerative ai engineer\b",
    r"\bai agent engineer\b",
    r"инженер.{0,10}искусственн.{0,10}интеллект",
    r"\bai[\s-]?инженер\b",
    r"разработчик.{0,10}искусственн.{0,10}интеллект"
],

    "ML Engineer": [
    r"\bmachine learning engineer\b",
    r"\bmachine learning developer\b",
    r"\bml[\s-]?engineer\b",
    r"\bml[\s-]?developer\b",
    r"\bml[\s-]?инженер\b",
    r"\bml[\s-]?разработчик\b",
    r"\bcomputer vision engineer\b",
    r"\bcv engineer\b",
    r"инженер.{0,10}машинн.{0,10}обуч",
    r"разработчик.{0,10}машинн.{0,10}обуч"
],

    "Data Scientist": [
        r"\bdata scientist\b",
        r"\bsenior data scientist\b",
        r"\bjunior data scientist\b",
        r"\bmiddle data scientist\b",
        r"\bdata science specialist\b",
        r"\bdata science scientist\b",
        r"дата[\s-]?сайентист",
        r"специалист.{0,10}data science"
    ],

    "Data Engineer": [
    r"\bdata engineer\b",
    r"\bdata platform engineer\b",
    r"инженер.{0,5}данных",
    r"\bdwh engineer\b",
    r"\betl engineer\b",
    r"\betl developer\b"
],

    "BI Analyst": [
    r"\bbi[\s-]?analyst\b",
    r"\bbi[\s-]?аналитик\b",
    r"\bbusiness intelligence analyst\b",
    r"\bpower bi analyst\b",
    r"\bbi developer\b",
    r"\bpower bi developer\b",
    r"\bbi[\s-]?разработчик\b",
    r"бизнес[\s-]?аналитик.{0,10}\bbi\b"
],

    "Data Analyst": [
    r"\bdata analyst\b",
    r"аналитик.{0,5}данных",
    r"\bproduct analyst\b",
    r"продуктов.{0,5}аналитик",
    r"\bdata analytics analyst\b"
],

    "Python Developer": [
        r"\bpython developer\b",
        r"\bpython engineer\b",
        r"\bpython backend developer\b",
        r"\bbackend python developer\b",
        r"python[\s-]?разработчик",
        r"разработчик.{0,5}python",
        r"python[\s-]?программист"
    ]
}
```

```python
def classify_role_v2(title):

    if not isinstance(title, str):
        return "Other"

    title = title.lower().strip()

    for role, patterns in ROLE_PATTERNS.items():

        for pattern in patterns:

            if re.search(pattern, title):
                return role

    return "Other"
```

```python
df["role"] = (
    df["title"]
    .apply(classify_role_v2)
)
```
